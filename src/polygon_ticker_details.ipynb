## Look up ticker details

```python
from dataclasses import asdict

ticker = "SPCX"

details = polygon_fetch_ticker_details(ticker)

# TickerDetails is a dataclass, so convert it to a dict, flatten the nested
# address/branding fields into dotted keys, and show one row per field
details_df = pd.json_normalize(asdict(details)).T.rename(columns={0: ticker}).dropna()
details_df.index.name = "Field"

# Show long fields (e.g. description) in full
with pd.option_context("display.max_colwidth", None):
    display(details_df)
```
