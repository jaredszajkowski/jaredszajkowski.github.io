 ## Python Imports

```python
# Standard Library
import os
import sys
import warnings

from pathlib import Path

# Data Handling
import numpy as np
import pandas as pd

# Display
from IPython.display import display

# Suppress warnings
warnings.filterwarnings("ignore")

# Add src/ to the path so settings.py and the helper modules can be imported
src_directory = str(Path.cwd().parents[2] / "src")
if src_directory not in sys.path:
    sys.path.append(src_directory)

from settings import config

DATA_DIR = config("DATA_DIR")  # keep only the directories this post uses
```

## Python Functions

Here are the functions needed for this project:

* [load_data](/posts/reusable-extensible-python-functions-financial-data-analysis/#load_data): Load data from a CSV, Excel, or Pickle file into a pandas DataFrame.

```python
from load_data import load_data
```

## Set Variables

```python
ticker = "IWM"
source = "Polygon"
asset_class = "Equities"
timeframe = "hour"
filter_date = "2026-06-01"
```

## Load Data From Pickle File

```python
df = load_data(
    base_directory=DATA_DIR,
    ticker=f"{ticker}",
    source=source,
    asset_class=asset_class,
    timeframe=timeframe,
    file_format="pickle",
)

display(df)
```

```python
display(df[df["Date"] > filter_date])
```

## Filter DataFrame

```python
df = df[df["Date"] > filter_date].reset_index(drop=True)
display(df)
```

## Export

```python
# Export the combined DataFrame to a pickle file
df.to_pickle(DATA_DIR / source / asset_class / timeframe / f"{ticker}.pkl")
```
